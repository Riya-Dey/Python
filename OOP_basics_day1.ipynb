OOP -> Object Oriented Programming --> way to organize code using Classes and Objects

Key Concepts -
- Class - Blueprint for creating objects
- Inheritance - Child class extends parent class (reuses code)
- Polymorphism - Child classes override parent methods -> same method name, different behavior
- Object: A thing created in memory that has data and behavior.
- Instance: An object that is created from a particular class.
- Attributes - variables inside class
- Methods - Functions inside class


```python
#Constructor - in python constructor is a special method(__init__()) that is automatically called when object is created, to initialize its data
class Student1:
  def __init__(self, name):         #__init__() --> initialize the object
    self.name = name              #self --> refers to the current object

s1 = Student1("Rahul")
print(s1.name)
```

```python
##Classes & Objects
#Simple Class definition
class store:                    #Blueprint for a store

  def __init__(self, store_id, location):
    self.store_id = store_id                    #Attribute
    self.location = location                    #Attribute

#Create objects (instances)
store1 = store(101, "Mysore")
store2 = store(102, "Hyderabad")

#Access attributes
print(store1.store_id)
print(store1.location)
print(store2.store_id)
print(store2.location)
```

```python
#Method - Function defined inside a class that performs an action on an object
#Difference between Method & Function -->
                    #Function -> Defined outside of a class
                    #Method -> Defined inside of a class

#Example of Method
class Student2:
  def study(self):            #defining 'study' method
    print("I am studying")

S2 = Student2()             #creating an object
S2.study()
```

```python
#Example of Method with Constructor
class Student3:

  def __init__(self,name):    #Constructor: sets initial data
    self.name = name        #self store name inside this object

  def study(self):            #defining 'study' method
    print(self.name, "is studying")

S3 = Student3("Riya")      #creating an object
S4 = Student3("Ayra")
S3.study()
S4.study()
```

Class vs Instance variables
- class variable:
  - shared by all objects
  - defined at class level (not in __init__())
  - same for all instances
- Instance variable:
  - Unique to each object
  - Defined in __init__() as self.variable
  - different for each instance

```python
class Student4:
    school = "XYZ School"          # Class variable: shared by everyone

    def __init__(self, name, age):
        self.name = name            # Instance variable: different for each object
        self.age = age              # Instance variable: different for each object

    def show(self):
        print(self.name, self.age, self.school)


s5 = Student4("Riya", 25)
s6 = Student4("Ayra", 27)

s5.show()
s6.show()
```

```python
#Inheritance - child(derived) class inherits the properties and methods of parent(base) class

class Animal:                 #parent class
  def eat(self):
    print("Animal eats")

class Cat(Animal):           #Child class
  def meow(self):
    print("Cat meows")

c = Cat()
c.eat()                   #Inherited from Animal
c.meow()                #Cat's own method
```

```python
#super() - access parent class method()

class Animal:                         # Parent class

    def __init__(self, name):
        self.name = name

    def eat(self):
        print(self.name, "eats")

class Dog(Animal):                    # Child class

    def __init__(self, name, breed):
        super().__init__(name)        # Call Parent's __init__()
        self.breed = breed            # Dog's own attribute

    def bark(self):
        print(self.name, "barks")

d = Dog("Tommy", "Labrador")

d.eat()                               # Inherited method
d.bark()                              # Dog's own method
print(d.breed)
```

```python
#method overriding

class Animal:                         # Parent class

    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):                    # Child class

    def sound(self):                  # Override parent's sound()
        print("Dog barks")

class Cat(Animal):                    # Another child class

    def sound(self):                  # Override parent's sound()
        print("Cat meows")

a = Animal()
d = Dog()
c = Cat()

a.sound()                             # Parent's method
d.sound()                             # Dog's overridden method
c.sound()                             # Cat's overridden method
```

```python
#Multilevel inheritance
#Grandparent -> Parent -> Child

class Animal:                         # Grandparent
    def eat(self):
        print("Animal eats")

class Dog(Animal):                    # Parent
    def bark(self):
        print("Dog barks")

class Puppy(Dog):                     # Child
    def play(self):
        print("Puppy plays")

p = Puppy()

p.eat()       # Inherited from Animal
p.bark()      # Inherited from Dog
p.play()      # Puppy’s own method
```
