```python
##Shallow Copy vs Deep Copy
# A shallow copy creates a new outer object, but the nested objects inside it are still shared with the original.
# A deep copy creates a new object and also creates separate copies of all nested objects.
```

```python
#Shallow copy

import copy

original = [[1, 2], [3, 4]]

# Create a shallow copy
shallow = copy.copy(original)

# Modify the nested list
shallow[0].append(5)

print(original)  # [[1, 2, 5], [3, 4]]
print(shallow)   # [[1, 2, 5], [3, 4]]
```

```python
#Deep Copy

import copy

original = [[1, 2], [3, 4]]

# Create a deep copy
deep = copy.deepcopy(original)

# Modify the nested list
deep[0].append(5)

print(original)  # [[1, 2], [3, 4]]
print(deep)      # [[1, 2, 5], [3, 4]]
```

```python
##Shallow Copy & Deep Copy

import copy

original = [[1, 2], [3, 4]]

shallow = copy.copy(original)
deep = copy.deepcopy(original)

# Change shallow copy
shallow[0].append(5)

# Change deep copy
deep[1].append(6)

print(original)
# [[1, 2, 5], [3, 4]]

print(shallow)
# [[1, 2, 5], [3, 4]]

print(deep)
# [[1, 2], [3, 4, 6]]
```
