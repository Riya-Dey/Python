```python
##Generator -> a special type of iterator that produces values one at a time, instead of creating and storing all values in memory at once.
  #The easiest way to create a generator is by using the yield keyword inside a function.

def numbers():
    yield 1
    yield 2

result = numbers()
print(result)
```

```python
#First Generator Example
def numbers():
    yield 1
    yield 2
    yield 3


generator = numbers()

print(next(generator))            #next() asks an iterator or generator to give next value
print(next(generator))
print(next(generator))
```

Yeild

yield is similar to return, but it behaves differently.

When Python reaches return:
- function starts
     ↓
- return 10
     ↓
- function ends


When Python reaches yield:
- function starts
     ↓
- yield 10
     ↓
- function pauses
     ↓
- value is given to caller



```python
#yield vs return

#return
def example():
    return 1
    return 2
    return 3              #can be returned,function ends after the first return.

#yeild
def example():
    yield 1
    yield 2
    yield 3               #one at a time.
```
