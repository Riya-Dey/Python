Importance of error handling ->
Data pipelines can fail for the below reasons
*   Missing files
*   Invalid data
*   Network timeouts
*   Wrong data types
*   Division by zero
With error handling - Program handles error gracefully, continues running

```python
##Basic try-except

#Simple try-except
'''Syntax:
try:
  code that might fall
except:
  code to run if error occurs'''

#Example1
try:
  age = int("not a number")
except:
  print("Error: Age must be a number")

#Example2
try:
  num1 = int("string")
except:
  print("Error: only numbers can be converted from string to integer")
```

Catching specific exceptions ->
Common exception types -
- ValueError -> Wrong value type --> int("Hello")
- TypeError -> Wrong data type --> "string" + 5
- KeyError -> Dict key not found --> dict["missing"]
- IndexError -> List index out of range --> [1,2,3,4][7]
- FileNotFoundError -> File doesn't exixt --> open(missing.txt)
- ZeroDivisionError -> Divide by zero --> 10/0
- AttributeError -> Object has no attribute --> obj.missing_attr

```python
#Example1
try:
  result1 = 10/0
except ZeroDivisionError:
  print("Error: cannot divide by zero")

#Example2
try:
  result2 = 5 + 2 + 'seven'
except TypeError:
  print("Error: cannot add string with numbers")
```

```python
#Multiple except caluse

try:
  file1 = open("missing.txt")
  num2 = int("string")
  result3 = 10/0
except FileNotFoundError:
  print("Error: file is missing")
except ValueError:
  print("Error: must be a number")
except ZeroDivisionError:
  print("Error: cannot divide by zero")
except Exception as e:                #catch all other errors
  print(f"Unexpected error: {e}")
#output - only first matching exception will run
```

```python
##try-except-finally
#Finally - Code in finally block always runs, even if error occurs.
#Used for cleanup(close files, release resources, etc.)
'''
Syntax:
try:
  code that might fall
except:
  handle error
finally:
  cleanup code(always runs)
  '''

#example1
try:
  file2 = open("missing_data.csv","r")
  data = file2.read()
except FileNotFoundError:
  print("Error: file is missing")
finally:
  print("File doesn't exist")
```

```python
##try-except-else
#Code in else block runs only if no error occurs in try block

#SImple try-except-else
try:
  age = int(input("Enter your age: "))
except ValueError:
  print("Error: age must be a number")
else:
  print(f"Your age is: {age}")
```

```python
##Raise - throwing exceptions
#Explicitly throw an error to signal a problem & stops execution unless caught with try-except block

#Raising exceptions
def validate_age(age):
  if age<0:
    raise ValueError("Age cannot be negative")
  if age>150:
    raise ValueError("Age is unrealistic")
  return True

try:
  validate_age(-2)
except ValueError as e:
  print(f"Error: {e}")
```

```python
##Custom exceptions
#To define exception classes for domain-specific errors, which makes errors meaningful & easier to debug

#Example
#Define custom exception
class DataValidationError(Exception):   #Raised when data validation failed
  pass

def validate_data(data):
  if not data:
    raise DataValidationError("Data is invalid")
  return True

try:
  validate_data("")
except DataValidationError as e:
  print(f"Data Validation Error: {e}")
```
