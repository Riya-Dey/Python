```python
##Control Flow - IF-ELSE Statements

#Simple IF
age = 17
if age >= 18:
  print("You are eligible for driver's licence")        #if the condition check fails, it'll return nothing
```

```python
#IF-ELSE
age =21
if age >= 18:
  print("You are eligible to apply for driver's license")
else:
  print("You are not eligible to apply for driver's license")
```

```python
#IF-ELIF
traffic_light = "green"
if traffic_light == 'green':
  print("you are free to go")
elif traffic_light == 'yellow':
  print("you need to slow down")
elif traffic_light == 'red':
  print("you need to stop")
```

```python
#IF-ELIF-ELSE
marks = 75

if marks >= 90:
  print("Grade A")
elif marks >= 80:
  print("Grade B")
elif marks >= 70:
  print("Grade C")
elif marks >= 65:
  print("Grade D")
else:
  print("Grade F")
```

```python
#Nested IF-ELSE
age = 18
income = 5000

if age >= 18:
  if income >= 30000:
    print("Eligible for loan")
  else:
    print("have too low income to be eligible for loan")
else:
  print("Too young to be eligible for loan")
```

```python
#Multiple Conditions
age = 45
citizenship = False

if age >= 18 and citizenship == True:
  print("Eligible to vote")
else:
  print("Not eligible to vote")
```

```python
#Ternary Operator
result = 75
status = "Pass" if result >= 65 else "Fail"
print(status)
```

```python
##Control Flow - FOR Loop

#Simple FOR Loop
for i in range(3):
  print("Numbers starting from 0, excluding last number:", i)

for i in range(1,5):
  print("Nmbers in range, excluding last number", i)
for i in range(0,11,2):
  print("printing numbers in a range skipped by two decimals", i)
```

```python
#FOR Loop - iterate over List
List1 = [1,5,'l',3.9]
for i in List1:
  print("iterate over List",i)

#FOR Loop - iterate over String
str1 = 'Python'
for i in str1:
  print("iterate over String",i)

#FOR Loop - iterate over Dictionary
dict1 = {'Fruit':'Coconut', 'Flower':'Lily', 'Season':'Summer'}
for i in dict1:
  print("iterate over Dictionary",i)
for key, value in dict1.items():
  print(f"{key} : {value}")                 #Output - key : value
```

```python
#Nested FOR Loop

for i in range(5):
  for j in range(3):
    print(f"({i}, {j})", end = " ")
  print()
```

```python
##Control Flow - WHILE Loop

#Simple WHILE Loop
num1 = 0
while num1 <= 5:
  print(num1)
  num1 +=1
```

```python
##Control Flow - Break Statement

#while loop
num2 = 0
while num2 <= 10:
  print(num2)
  if num2 == 5:
    break                 #Break - stops entire loop
  num2 += 1
```

```python
#FOR loop

for i in range(5):
  if i == 3:
    break
  print(i)
```

```python
##Control Flow - Continue Statement

#while loop
num3 = 0
while num3 < 10:
  num3 += 1
  if num3 % 2 == 0:
    continue                  #Continue - skips current iteration in loop
  print(num3)
```

```python
#FOR loop

for i in range(10):
  if i % 2 ==0:
    continue
  print(i)
```
