```python
##Collections

#List
list1 = [7,4,4,5,3,'X',5,2.7]
print(f"A list with random values and data types: {list1}")
list1_len = len(list1)
print(f"The length of the list is: {list1_len}")
list1.append(10)
print(f"after appending single integer value in list1: {list1}")
list1.extend([6,"Z"])                                    #l st.extend() takes exactly one argument
print(f"after adding multiple values in list1: {list1}")
list1.insert(6,"Y")                                       # insert expects 2 arguments
print(f"after adding single value in specific position: {list1}")
list1.pop()
print(f"after removing last item: {list1}")
list1.remove(5)                                         # remove the first found value
print(f"after removing a specific item: {list1}")
list1.reverse()
print(f"after reversing the list, list1: {list1}")
7 in list1                                              # searching for a specific item in list1
```

```python
#Tuple
tuple1 = (7, 4, 4, 5, 3, 'X', 5, 2.7)
print(f"A tuple with random values and data types: {tuple1}")
len_tup = len(tuple1)
print(f"The length of tuple1 is: {len_tup}")
count_tup = tuple1.count(4)
print(f"Count of specific item(4) in tuple is: {count_tup}")
ind_tup = tuple1.index(4)                                   #Only shows the result for 1st found item
print(f"Index of specific item(4) in tuple is: {ind_tup}")
```

```python
#Set
empty_set = set()
print(type(empty_set))
print(f"Creating empty set: {empty_set}")
set1 = {1,5,5,"BMW",4.2}
print(f"A set with random values: {set1}")
set1.add(6)
print(f"after adding single value in set: {set1}")
set1.update([5,"xyz",'abc',2])
print(f"after adding multiple values in set: {set1}")
set1.remove('abc')                                  # if not found, it'll throw error
print(f"after removing existing item: {set1}")
set1.discard('abc')                                 # if not found, it'll not throw error
print(f"after removing non-existing item: {set1}")
set1.pop()
print(f"after removing an arbitrary item: {set1}")

#Set Operations
set2 = {4,6,5}
set3 = {4,2,3,1}

union_set = set2.union(set3)
print(f"Union operation on 2 sets: {union_set}")
intersection_set = set2.intersection(set3)
print(f"Intersection operation on 2 sets: {intersection_set}")
difference_set = set2.difference(set3)
print(f"Difference_set operation on 2 sets: {difference_set}")
```

```python
#Dictionary
empty_dict1 = {}
empty_dict2 = dict()
print(type(empty_dict1),type(empty_dict2))

dict1 = {'name' : 'Riya', 'position':'Data Engineer', 'age':25, 'city' :'Hyderabad'}
print(f"creating dictionary: {dict1}")
print(f"accessing value from key: {dict1['name']}")
print(f"accessing value from key: {dict1.get('age')}")
print(f"accessing value from key: {dict1.get('height')}")   #return none if can't find key in dictionary
dict1['age'] = 26
print(f"modifying value for a key: {dict1}")
dict1['gender'] = 'female'
print(f"adding key-value pair: {dict1}")
del dict1['city']
print(f"deleting specific key-value pair: {dict1}")         # del dict1[key] removes the key and does not return anything
dict1.pop('age')
print(f"deleting specific key-value pair: {dict1}")         # dict1.pop(key) removes the key and returns its value
print(f"fetching all the keys: {dict1.keys()}")
print(f"fetching all the values: {dict1.values()}")
print(f"fetching all the items: {dict1.items()}")
```
