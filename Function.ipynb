```python
#Function - A reusable code block that performs specific task. Reduces repetition & improves code readability

##Basic Functions

#Simple Function
def greet():
  print ("Hello! It's a simple function format")
greet()

#Function with Parameters & Arguments
def sum(a,b):                                 #a,b are parameters, working as placeholders
  print(f"Sum of 2 given parameters: {a+b}")
sum(6,7)                                      #6,7 are arguments, the actual value when calling the function

#Function with Return value
def add(x,y):
  return x*y
print(f"Function with return value: {add(2,5)}")

#Function with multiple Return values
def calc(y,z):
  sum_val = y+z
  product_val = y*z
  return sum_val, product_val

sum_val1, product_val1 = calc(2,3)
print(f"Function with multiple return values: {sum_val1, product_val1}")

#Function with positional arguments

def identity(name, age, city):
  print(f"User's identity: {name}, {age}, {city}")
identity('Riya', 25, 'Hyderabad')                       #need to give arguments order-wise

#Function with named arguments

def identity(name, age, city):
  print(f"User's identity: {name}, {age}, {city}")
identity(name = 'Riya', city = 'Hyderabad', age = 25)   #no need to give arguments order-wise
```

```python
#Scope - The area where a variable is accessible. Python has 4 scopes (LEGB):
          # Local (L) - Inside function
          # Enclosing (E) - In nested function
          # Global (G) - Outside all functions
          # Built-in (B) - Python built-in (len, print etc.)

#Scope priority (LEGB): Local (L) > Enclosing (E) > Global (G) > Built-in (B)
```

```python
#Local Scope
def greet():
  name = 'Riya'           #Local variable
  print(f"Hello {name}")
greet()
#print(name)       # will give error outside of the function --> NameError: name 'name' is not defined
```

```python
#Global Scope
g = "Global variable"

def func1():
  print(f"This is: {g}")
func1()
print(g)              #can access outside the function
```

```python
#Local Scope & Global Scope in same function

#Without Global Keyword
x = 5
def num1():
  x = 10          #Local variable shadows Global variable
  print(x)
num1()          #Output - Local variable
print(x)        #Global variable remains unchanged


#With Global Keyword
x = 5
def num2():
  global x          #Global keyword modifies the value of the variable
  x = 10
  print(x)
num2()          #Output - Local variable
print(x)        #Global variable's value is changed
```

```python
#Nested functions (Enclosing Scope)

#Without Nonlocal Keyword
def outer():
  x = 10        #Enclosing Scope

  def inner():
    print(x)    #can access outer function's variable
  inner()

outer()
```

```python
#With Nonlocal Keyword
def outer():
  x = 10        #Enclosing Scope

  def inner():
    nonlocal x
    x = 5    #can modify parent's variable
    print(x)
  inner()

outer()
```

```python
#Default Parameter - Parameters with pre-set values

def num3(z=7):
  print(z)
num3()          #Default value is 7
num3(10)        #Modified value is 10
```

```python
##*args - non-keyword arguments (variable number of positional arguments)
          #allows function to accept any number of arguments
          #*args is tuple
```

```python
#Simple *args

def print_num1(*args):
  print(args)           #Tuple of all arguments

print_num1(5,6,8,5)   #Output - Tuple


def print_num2(*args):
  for num in args:
    print(num)

print_num2(5,6,8,5)   #Output - numbers
```

```python
##**kwargs - keyword arguments (variable number of key-value arguments)
            #allows function to accept any number of named arguments
            #**kwargs is a dictionary
```

```python
#Simple **kwargs

def print_id(**kwargs):
  print(kwargs)               #Dictionary of all keyword arguments
  for key, value in kwargs.items():
    print(f"{key} : {value}")
print_id(name='Riya', age=25, city='Hyderabad')
```
