Decorator - wrapper around a function

a function that modifies or extends the behavior of another function without changing its actual code.

Use cases ->
- Logging — record which functions were called.
- Timing — measure execution time.
- Authentication/authorization — check whether a user has permission.
- Validation — check function inputs.
- Caching — reuse previously calculated results.
- Error handling — provide common error-handling behavior.


```python
#Simple Decorator

def my_decorator(func):             #decorator function declaration

    def wrapper():                  # This wrapper function adds extra behavior
        print("Before the function runs")
        func()                      # Call the original function
        print("After the function runs")
    return wrapper

# Apply the decorator
@my_decorator
def say_hello():
    print("Hello!")

# Calling say_hello() actually calls wrapper()
say_hello()
```

```python
#Decorators with Arguments

def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("Function is starting...")
        result = func(*args, **kwargs)          # Pass all arguments to the original function
        print("Function has finished!")

        return result

    return wrapper

@my_decorator
def add(a, b):
    return a + b

result = add(10, 20)
print("Result:", result)
```

Magic Methods -> also called dunder methods ("double underscore" methods).

special methods in Python that start and end with double underscores (__).

Magic Method	Used for -->
- __init__()	-> Initialize an object
- __str__()	-> Control what print(object) displays
- __repr__()	-> Developer-friendly representation
- __len__()	-> Make len(object) work
- __add__()	-> Define behavior of object1 + object2
- __eq__()	-> Define behavior of object1 == object2
- __lt__()	-> Define behavior of object1 < object2

```python
##Magic methods

class Student:

    def __init__(self, name, marks):              # __init__ runs automatically when an object is created
        self.name = name
        self.marks = marks

    def __str__(self):                            # __str__ controls what print(object) displays
        return f"{self.name} - {self.marks} marks"

    def __len__(self):                            # __len__ controls what len(object) returns
        return len(self.name)

    def __eq__(self, other):                      # __eq__ controls the == operator
        return self.marks == other.marks

    def __lt__(self, other):                      # __lt__ controls the < operator
        return self.marks < other.marks

student1 = Student("Riya", 90)
student2 = Student("Bidisha", 90)
student3 = Student("Jaya", 75)

print(student1)                           # __str__()
print(len(student1))                      # __len__()
print(student1 == student2)              # __eq__()
print(student3 < student1)               # __lt__()
```
