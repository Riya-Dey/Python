```python
##Polymorphism - the same method/function name can behave differently depending on the object or data you use it with

```

```python
#Method Overriding - child claasses override the same method from the parent class

class Animal:
  def speak(self):
    print("Animal make sound")

class Cat(Animal):
  def speak(self):
    print("Cat meows")

class Dog(Animal):
  def speak(self):
    print("Dog barks")

cat = Cat()
dog = Dog()
cat.speak()
dog.speak()

#Output - Both Dog and Cat have the same method name speak(), but each behaves differently
```

```python
#Duck Typing - Duck typing means Python cares about what an object can do, not what type it is
#No need to check the object's class, simply call the method we need

class Dog:
  def speak(self):
    print("Woof!")

class Cat:
  def speak(self):
    print("Meow!")

class Robot:
  def speak(self):
    print("Hello!")

def sound(obj):
  obj.speak()

sound(Dog())
sound(Cat())
sound(Robot())

#sound() doesn't care whether the object is a Cat, Dog or Robot. It only expects the object to have a speak() method.
```

```python
#Operator Overloading - Polymorphism can also be used with operators such as +, -, ==, etc.

class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Number(self.value + other.value)

num1 = Number(10)
num2 = Number(20)
result = num1 + num2
print(result.value)
```

```python
#Function Polymorphism - same function works with different objects.

class Dog:
    def speak(self):
        return "Woof!"
class Cat:
    def speak(self):
        return "Meow!"

def animal_sound(animal):
    print(animal.speak())

dog = Dog()
cat = Cat()
animal_sound(dog)
animal_sound(cat)

#The animal_sound() function is written only once, but it works with multiple object types.
```
